```python
import pandas as pd
import requests

url = "https://geoportal.siata.gov.co/fastgeoapi/geodata/geographJson/1/pm25/86"
respuesta = requests.get(url)

if respuesta.status_code == 200:
    datos = respuesta.json()
    info = datos["info"]
    
    # Crear un DataFrame uniendo las horas con sus respectivos valores de PM2.5
    df = pd.DataFrame({
        "Hora": info["Horas"],
        "PM25": info["PM25_72H"]
    })
    
    # Ver las primeras filas de la tabla
    print(df.head())
    
    # Ejemplo analítico: Calcular el promedio real omitiendo los valores nulos
    promedio_pm25 = df["PM25"].mean()
    print(f"\nEl promedio real de las 72 horas es: {promedio_pm25:.2f} μg/m³")

import matplotlib.pyplot as plt

plt.plot(df["PM25"])
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import requests

url = "https://geoportal.siata.gov.co/fastgeoapi/geodata/geographJson/2/velocidad_viento_6h/198"
respuesta = requests.get(url)

if respuesta.status_code == 200:
    datos = respuesta.json()
    info = datos["info"]
    
    # Crear un DataFrame uniendo las horas con sus respectivos valores de PM2.5
    df = pd.DataFrame({
        "Tiempo": info["Tiempo"],
        "Viento": info["Viento"]
    })
    
    # Ver las primeras filas de la tabla
    print(df.head())
    
    # Ejemplo analítico: Calcular el promedio real omitiendo los valores nulos
    promedio_pm25 = df["Viento"].mean()
    print(f"\nEl promedio real de las 72 horas es: {promedio_pm25:.2f} μg/m³")


plt.plot(df["Viento"])
```
